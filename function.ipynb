6. Explain the concept of generators in Python and how they are defined.

> Generators are created using the yield keyword.

```python
#example

def numbers():
    yield 11
    yield 22
    yield 33
for n in numbers():
    print(n)

    #the yield statement pauses the function
    #also continues from the same point when the next value is requested
```

7. What are the advantages of using generators over regular functions?

Generators have some useful advantages:

> 1. They use less memory because they produce one value at a time.

> 2. They are useful when working with large amounts of data.

> 3. They do not need to store all the values in memory.

> 4. They are useful when values are needed one by one.

> 5. They are easy to create using the yield keyword.

> For example:
  if we need to generate a very large number of values using a generator can be more memory efficient than creating a large list.

10. What is the difference between map(), reduce(), and filter() functions in Python?

>  map(), reduce() and filter() are used to perform operations on collections of data but they work differently.


> The main difference is :

> map() - changes each element

> filter() - selects elements

> reduce() - combines elements into one results

```python
#map()
#it applies a function to every element

map(lambda x: x * 2, numbers)
```

```python
#filter()
#it selects only the elements that satisfy a condition

filter(lambda x: x % 2 == 0,numbers)
```

```python
#reduce()
#it combines the elements repeatedly which gives one final result and it is available from the functools module

from functools import reduce

reduce(lambda x, y: x + y, numbers)
```

the list.

> Sum of all even numbers in a list :

```python
def sum_even(numbers):
  total = 0

  for num in numbers:
    if num % 2 == 0:
      total += num
  return total

numbers = [10, 15, 20, 25, 30]


print("Su of even numbers:" , sum_even(numbers))
```

9. Write a Python program that uses map() to convert a list of temperatures from Celsius to Fahrenheit.

> Convert Celsius temperatures to Fahrenheit using map():
  
  
> formula: Fahrenheit = (Celsius * 9/5) + 32



```python
celsius = [0,10,20,30,40]

fahrenheit = list(map(lambda c: (c * 9/5)+ 32, celsius))


print("Celsius:", celsius)
print("Fahrenheit:", fahrenheit)
```
